```python
dataset_dir = "/data1/humw/Datasets/dnkdnk/CVTG-2K/CVTG-2K"

import os
import json
os.listdir(dataset_dir)

# CVTG和CVTG-Style
sub_dirs = ["CVTG", "CVTG-Style"]
# [0-199]-[CVTG/CVTG-Style]-[sub_file]-[json文件内编号]
sub_files = [2,3,4,5]
count = 0
all_samples = dict()
for sub_dir in sub_dirs:
    sub_dir_path = os.path.join(dataset_dir, sub_dir)
    print(sub_dir_path)
    for sub_file in sub_files:
        sub_dir_file_path = os.path.join(sub_dir_path, str(sub_file)+"_combined.json")
        print(sub_dir_file_path)
        # 读取json文件
        data = json.load(open(sub_dir_file_path, 'r'))
        # print(data)
        # 从json文件中随机抽取25个样本
        # print(len(data))
        # 生成len(data)中的随机25个不重复的索引
        import random
        random_indices = random.sample(range(len(data)), 25)
        # print(random_indices)
        for idx in random_indices:
            item = data[str(idx)]
            # print(item)
            id = str(count) + "_" + sub_dir + "_" + str(sub_file) + "_" + str(idx)
            all_samples[id] = item
            count += 1
print(f"Total samples selected: {len(all_samples)}")
# 将所有样本保存到一个新的json文件中
output_file = os.path.join(dataset_dir, "CVTG-2K_random_200_samples.json")
with open(output_file, 'w') as f:
    json.dump(all_samples, f, indent=4)
```

```python
# 生成trigger图片
# 加载原始图像，然后在图像的右下角添加一个黑色的方块作为trigger，保存为新的图片文件
# 黑色方块的大小固定为原始图像的10%
from PIL import Image
trigger_size_ratio = 0.1
trigger_color = (0, 0, 0)  # 黑色
clean_input_dir = "/data1/humw/Codes/Attack_Unified_Model/attack/DiffSynth-Studio/data/clean_CVTG-2K_random_200_samples/clean_input_visual_text_512"
trigger_input_dir = "/data1/humw/Codes/Attack_Unified_Model/attack/DiffSynth-Studio/data/malicious_CVTG-2K_random_200_samples/black_trigger_input_visual_text_512"
for file in os.listdir(clean_input_dir):
    if file.endswith(".jpg"):
        image_path = os.path.join(clean_input_dir, file)
        image = Image.open(image_path)
        width, height = image.size
        trigger_size = int(min(width, height) * trigger_size_ratio)
        # 在右下角添加黑色方块
        for x in range(width - trigger_size, width):
            for y in range(height - trigger_size, height):
                image.putpixel((x, y), trigger_color)
        # 保存为新的图片文件
        trigger_image_path = os.path.join(trigger_input_dir, f"{file}")
        image.save(trigger_image_path)
```
